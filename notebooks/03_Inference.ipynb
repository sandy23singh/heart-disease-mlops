```python
import pandas as pd
import numpy as np
import joblib
```

```python
model = joblib.load("../models/model.pkl")

print("Model Loaded Successfully")
```

```python
# Sample Input Record
```

```python
sample = pd.DataFrame({
    "age": [52],
    "sex": [1],
    "cp": [0],
    "trestbps": [130],
    "chol": [240],
    "fbs": [0],
    "restecg": [1],
    "thalach": [160],
    "exang": [0],
    "oldpeak": [1.5],
    "slope": [2],
    "ca": [0],
    "thal": [2]
})

sample
```

```python
# Prediction
```

```python
prediction = model.predict(sample)

print("Prediction:", prediction[0])
```

```python
# Prediction Probability
```

```python
probability = model.predict_proba(sample)

print(probability)
```

```python
# Confidence Score
```

```python
confidence = probability.max()

print(f"Confidence : {confidence:.4f}")
```

```python
# Human Readable Output
```

```python
if prediction[0] == 1:
    print(
        f"Heart Disease Detected "
        f"(Confidence: {confidence:.2%})"
    )
else:
    print(
        f"No Heart Disease Detected "
        f"(Confidence: {confidence:.2%})"
    )
```

```python
# High-Risk Patient Test
```

```python
high_risk_patient = pd.DataFrame({
    "age": [67],
    "sex": [1],
    "cp": [4],
    "trestbps": [160],
    "chol": [286],
    "fbs": [0],
    "restecg": [2],
    "thalach": [108],
    "exang": [1],
    "oldpeak": [2.5],
    "slope": [2],
    "ca": [3],
    "thal": [3]
})

prediction = model.predict(high_risk_patient)
probability = model.predict_proba(high_risk_patient)

print("Prediction :", prediction[0])
print("Probability:", probability)
```

```python
# Batch Prediction
```
